```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Set a consistent plotting style
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11
```

```python
# Examine dataset information
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Display summary statistics
display(df.describe())
```

```python
# Create a separate copy for cleaning
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Standardize shooting percentages
percentage_columns = ["fg_pct", "three_pct", "ft_pct"]

for column in percentage_columns:
    df_clean.loc[df_clean[column] > 1, column] = (
        df_clean.loc[df_clean[column] > 1, column] / 100
    )

# Display updated dataset information
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df_clean.duplicated().sum())
```

```python
# Save the cleaned dataset
df_clean.to_csv("metro-hoops-cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```

# SECTION I

# HISTOGRAM OF SCORE

```python
# Histogram showing the distribution of points per game

points = df_clean["points_per_game"].dropna()

plt.figure()

plt.hist(points, bins=15, edgecolor="black")

plt.axvline(
    points.mean(),
    linestyle="--",
    label=f"Mean: {points.mean():.2f}"
)

plt.title("Distribution of Points Per Game")
plt.xlabel("Points Per Game")
plt.ylabel("Number of Player-Season Records")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.annotate(
    f"Mean = {points.mean():.2f}",
    xy=(points.mean(), 0),
    xytext=(10, 15),
    textcoords="offset points"
)

plt.show()
```

# COUNT PLOT

```python
# Count player-season records by position

position_counts = df_clean["position"].value_counts()

plt.figure()

bars = plt.bar(
    position_counts.index,
    position_counts.values,
    edgecolor="black"
)

plt.title("Number of Player-Season Records by Position")
plt.xlabel("Player Position")
plt.ylabel("Number of Records")

# Add values above each bar
for bar in bars:
    height = bar.get_height()
    plt.annotate(
        str(int(height)),
        (bar.get_x() + bar.get_width() / 2, height),
        ha="center",
        xytext=(0, 5),
        textcoords="offset points"
    )

plt.grid(axis="y", alpha=0.3)
plt.show()
```

## EXPLANATION

The histogram displays the distribution of points scored per game by players in the dataset. It shows how frequently players fall within different scoring ranges. Most players score within the lower-to-middle point ranges, while fewer players average a high number of points per game. And the bar chart compares the number of players in each basketball position: Center, Forward, and Guard. It helps identify which positions are most represented in the dataset and shows how players are distributed across the three positions.

# GROUPBY BAR PLOT

```python
# Group data by position and calculate average statistics

position_averages = df_clean.groupby("position")[
    ["points_per_game", "rebounds_per_game", "assists_per_game"]
].mean()

# Create grouped bar chart

ax = position_averages.plot(
    kind="bar",
    figsize=(10, 6),
    edgecolor="black"
)

plt.title("Average Basketball Statistics by Position")
plt.xlabel("Player Position")
plt.ylabel("Average Per Game")

plt.xticks(rotation=0)
plt.legend(
    ["Points Per Game", "Rebounds Per Game", "Assists Per Game"]
)

# Annotate bars
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", padding=2, fontsize=8)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Explanation

The grouped bar chart compares the average points, rebounds, and assists per game for Centers, Forwards, and Guards. The results show that Centers have the highest average points and rebounds, while Guards have the highest average assists. This demonstrates how different basketball positions contribute to the game in different ways based on their statistical performance.

# SECTION II

```python
# Scatter plot of minutes vs points

plt.figure()

plt.scatter(
    df_clean["minutes_per_game"],
    df_clean["points_per_game"],
    color="purple",
    edgecolor="black",
    alpha=0.7,
    s=60
)

plt.title("Minutes Played vs. Points Per Game")
plt.xlabel("Minutes Per Game")
plt.ylabel("Points Per Game")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 6))

# Select key numerical columns
num_cols = ["age", "height_cm", "games", "minutes_per_game", "points_per_game", "rebounds_per_game", "assists_per_game"]
corr_matrix = df_clean[num_cols].corr()

# Plot heatmap
plt.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation Coefficient")

# Add labels
plt.xticks(range(len(num_cols)), num_cols, rotation=45, ha="right")
plt.yticks(range(len(num_cols)), num_cols)
plt.title("Correlation Heatmap of Player Statistics")

# Display correlation values inside the cells
for i in range(len(num_cols)):
    for j in range(len(num_cols)):
        plt.text(j, i, f"{corr_matrix.iloc[i, j]:.2f}",
                 ha="center", va="center", color="black", fontsize=9)

plt.tight_layout()
plt.show()
```

**Interpretation of the Scatter Plot**

The scatter plot illustrates the relationship between playing time (minutes_per_game) and scoring output (points_per_game) across player positions. Overall, it shows a strong, positive linear trend, meaning that players who play more minutes on the court consistently score higher points per game. Guards, forwards, and centers are well-distributed across the minutes spectrum, with elite scorers averaging over 20 points typically playing 30+ minutes regardless of position. Most player records cluster in the middle ranges (15–30 minutes, 5–15 points), while top-performing outliers occupy the upper-right corner.

**Interpretation of the Heatmap**

The correlation heatmap displays the numerical relationships between player statistics on a scale from -1 to 1. The strongest relationship in the dataset is between minutes_per_game and points_per_game (≈0.90), confirming that playing time is the primary driver of offensive output. Additionally, player height (height_cm) shows a positive correlation with rebounds (rebounds_per_game), indicating that taller players naturally secure more rebounds near the basket. Most other variables exhibit weak correlations, showing that metrics like age and total games played operate independently.
